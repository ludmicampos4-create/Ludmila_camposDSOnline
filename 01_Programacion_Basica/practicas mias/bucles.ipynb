```python
frase="En un lugar de La Mancha, de cuyo nombre no quiero acordarme"
contador=0
for letra in frase.lower(): #para que cuente desde el primero en la lista
    if letra == "m": #se pone el == para que busque las letras que sean iguales a m 
        contador +=1 #contador es igual a contador mas 1, para que cuente las veces que aparece la letra m
        print(f"la letra m aparece; {contador}")
        
```
