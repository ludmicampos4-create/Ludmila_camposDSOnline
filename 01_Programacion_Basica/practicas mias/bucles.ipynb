```python
class Tienda:
    """
    Clase que representa una tienda de electrodomésticos con atributos compartidos, de instancia y métodos para gestionar ventas y proyecciones de marketing.
    """
    tipo= "electrodomesticos"
    abierta= True
    def __init__(self, nombre, direccion, num_empleados, ventas_3_meses):
        self.nombre = nombre
        self.direccion = direccion
        self.num_empleados = num_empleados
        self.ventas_3_meses = ventas_3_meses
    def total_ventas(self):
        return sum(self.ventas_3_meses)
    def ventas_por_empleado(self):
        if self.num_empleados ==0:
            return 0
        return sum(self.ventas_3_meses)/self.num_empleados
    def ventas_ultimo_mes(self):
        return self.ventas_3_meses[-1]
    def proyeccion_marketing(self,proyeccion_marketing_x):
        if proyeccion_marketing_x <1000:
            self.ventas_3_meses= [v * 1.2 for v in self.ventas_3_meses]
        else:
            self.ventas_3_meses=[v * 1.5 for v in self.ventas_3_meses]
            return self.ventas_3_meses
tienda1= Tienda("electrosur", "calle mayor 123", 5, [1000, 1500, 2000])  
tienda2= Tienda("electrocentro", "avenida de la consitucion 45", 8, [2000, 2500, 3000])
tienda3= Tienda("Electronorte", "calle el sol 67", 10, [3000, 3500, 4000])

print("---COMPROBACION TIENDA 1---")
print("tipo:",tienda1.tipo)
print("abierta:", tienda1.abierta)
print("nombre y direccion:", tienda1.nombre, tienda1.direccion)
print("total ventas (3 meses):", tienda1.total_ventas())
print("ventas por empleado:", tienda1.ventas_por_empleado())
print("venta del ultimo mes:", tienda1.ventas_ultimo_mes())
print("proyeccion tras invertir 1200 en marketing:", tienda1.proyeccion_marketing(1200))

todas_las_tiendas= [tienda1, tienda2, tienda3]
total_ultimo_mes= 0
for tienda in todas_las_tiendas:
    total_ultimo_mes += tienda.ventas_ultimo_mes()
print(f"\ntotal ventas del ultimo mes de todas las tiendas: {total_ultimo_mes}---")
print("\n---tiendas ubicadas en una avenida---")
for tienda in todas_las_tiendas:
    if "avenida" in tienda.direccion.lower():
        print(f"{tienda.nombre} esta ubicada en una avenida")
```
